```python
import mpmath as mp
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
%matplotlib ipympl
```

```python
def calc_covertness(na, nb, m):
    nt = mp.floor(m*mp.log( (na+1)/(nb+1)) /mp.log( na*(nb+1)/(nb*(na+1))))
    outside = 2/( (na+1)**m*(nb+1)**m) *mp.binomial(nt+m-1,nt)
    def inner(n1,n2):
        return (n1/(n1+1))**nt *(n2+1)**m *(-1 + mp.hyp2f1(1,m+nt,1+nt,n1/(n1+1)))
    pos = inner(nb,na)
    neg = inner(na,nb)
    return outside*(pos-neg)
vec_calc_covertness = np.vectorize(calc_covertness)
```

```python
m = 300
nas = np.logspace(-8,0,100)
nl = np.logspace(-8,0,100)
nagrid,nlgrid = np.meshgrid(nas,nl)
nbgrid = nagrid+nlgrid
```

```python
covertness = vec_calc_covertness(nagrid,nbgrid,m)
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(nagrid,nlgrid,covertness)
ax.set_xlabel('Noise')
ax.set_ylabel('Signal')
ax.set_zlabel('Detectability')
plt.show()
```

```python
miss_prob = .5 -.25*covertness
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(nagrid,nlgrid,miss_prob)
ax.set_xlabel('Noise')
ax.set_ylabel('Signal')
ax.set_zlabel('Miss prob')
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(nagrid,nlgrid,miss_prob)
ax.set_xlabel('Noise')
ax.set_ylabel('Signal')
ax.set_zlabel('Miss prob')
ax.set_xscale('log')
ax.set_yscale('log')
plt.show()
```

```python
def calc_var(eta,n):
    sreta = np.sqrt(eta)
    A = (1-sreta)*((1-sreta)*(1+n)*(1-2*sreta+3*(1-2*sreta +eta)*(1+n)*n) +2*(1-eta)*(1+n)*(1+(1-sreta)*n) +(1+(1-sreta)*n)**2*(2+(2-n)*sreta))
    B = np.sqrt(n) *(2/sreta +2*(1/sreta-1)*n-1)
    return np.sqrt(A)/B
```

```python
n = 1000
etavals = np.logspace(-8,0,n)
nvals = np.logspace(-8,1,n)
etas,ns = np.meshgrid(etavals,nvals)
variance = calc_var(etas,ns)
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(etas,ns,variance)
ax.set_xlabel('eta')
ax.set_ylabel('n')
ax.set_zlabel('variance')
ax.set_xscale('log')
ax.set_yscale('log')
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(etas,ns,variance*np.sqrt(ns)/np.sqrt(etas))
ax.set_xlabel('eta')
ax.set_ylabel('n')
ax.set_zlabel('variance, scaled')
ax.set_xscale('log')
ax.set_yscale('log')
plt.show()
```
